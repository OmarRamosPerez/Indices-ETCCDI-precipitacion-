## Ploteo en malla — `ETCCDI_precip_plot_grid`

Genera mapas geoespaciales a partir de los archivos NetCDF calculados en la sección 1.

Hay dos funciones:
- `Plot_netcdf_1_tiempo` → Mapa de **un solo campo** (Distribución espacial del índice)
- `plot_netcdf_n_tiempos` → Mapa de **tendencias** (Mann-Kendall + Theil-Sen en toda la grilla)

## Archivos NetCDF (N tiempos)
**********************************
Nota: Los archivos NetCDF de los índices de la categoría "Duración" \
      Son de N tiempos 
**********************************

Las funciones generarán un mapa de **tendencias** 
(Theil-Sen en toda la malla y celdas que tienen significancia (p<0.05))

- Entrada: Un archivo .nc \
    Dimensiones: `(time , lat, lon)` 

- Ruta de salida: \
    Figura → Donde se guardará mapa de tendencias para una serie de campos anuales

    Estadísticos → NetCDF donde se guardará los campos de tendencia
