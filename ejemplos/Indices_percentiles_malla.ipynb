## Índices ETCCDI de Precipitación —
# Percentiles

| ID | Nombre | Definición | Unidades |
|-----------|----|--------|------------|
|**R95p**| Días muy húmedos | Precip. total anual o estacional acumulada en días > percentil 95 | mm |
|**R99p**| Días extremadamente húmedos | Precip. total anual o estacional acumulada en días > percentil 99 | mm |

## Datos en Malla
Las funciones para calcular los índices Categoría "Percentiles"  
Trabajan con datos de precipitación en formato NetCDF

- Ruta de salida:  Donde se guardará el archivo NetCDF con el índice calculado. 
    
    Formato esperado del archivo de salida: NetCDF con variable (índice calculado) 

    Dimensiones: (time, lat, lon)

## Período base
Las funciones requieren definir el período base para calcular el percentil
de acuerdo al período de tus datos pueden ser:
- 1961-1990
- 1981-2010
- 1991-2020
