## Ploteo en malla — `ETCCDI_precip_plot_grid`

Genera mapas geoespaciales a partir de los archivos NetCDF calculados en la sección 1.

Hay dos funciones:
- `Plot_netcdf_1_tiempo` → Mapa de **un solo campo** (Distribución espacial del índice)
- `plot_netcdf_n_tiempos` → Mapa de **tendencias** (Mann-Kendall + Theil-Sen en toda la grilla)

## Archivos NetCDF (1 tiempo)
**********************************
Nota: Los archivos NetCDF de los índices de la categoría "Frecuencia" \
      Son de 1 tiempo
**********************************

Las funciones generan "Un Mapa de un solo campo" a partir de los archivos .nc generados en la Notebook % run Indices_frecuencia_in_situ.ipynb.

- Entrada: Un archivo .nc \
    Dimensiones: `(time , lat, lon)`   

- Ruta de salida: Donde se guardará la figura del índice correspondiente.
