## Datos en malla — `ETCCDI_precip_malla`

Las funciones para calcular los índices Categoria Frecuencia requieren:

- Entrada: Un archivo NetCDF de precipitación diaria 
**Formato esperado del archivo de entrada:**
    NetCDF con variable de precipitación diaria en `mm/día`
    Dimensiones: `(time, lat, lon)`

- Ruta de salida: Donde se guardará el archivo NetCDF con el índice calculado.
 **Formato esperado del archivo de entrada:**
    NetCDF con variable (índice calculado)
    Dimensiones: `(time, lat, lon)`
