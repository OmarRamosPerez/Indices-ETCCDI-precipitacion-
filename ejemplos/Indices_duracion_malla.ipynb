## Datos en malla — `ETCCDI_precip_malla`

Las funciones para calcular los índices Categoria "Duración" requieren:

- Entrada: Un archivo NetCDF de precipitación diaria 
**Formato esperado del archivo de entrada:**
    NetCDF con variable de precipitación diaria en `mm/día`
    Dimensiones: `(time, lat, lon)`

- Ruta de salida: Donde se guardará el archivo NetCDF con el índice calculado.
 **Formato esperado del archivo de entrada:**
    NetCDF con variable (índice calculado)
    Dimensiones: `(time, lat, lon)`


## Índice CDD - Ejemplo

El archivo de entrada es: CanESM2_CanRCM4_2006-2023_day.nc

```python
# ── Rutas de entrada/salida (ajusta según tus directorios) ─────────────
ARCHIVO_PR = '/home/omar/Documents/Indices_Precip/Datos_entrada/CanESM2_CanRCM4_2006-2023_day.nc'   # NetCDF de precipitación diaria
ARCHIVO_SALIDA = '/home/omar/Documents/Indices_Precip/Indices_Duracion/CDD_CanESM2_CanRCM4.nc'
```

## Índice CDD (Días secos (PR < 1 mm) consecutivos)

```python
# CDD — Días secos consecutivos (PR < 1 mm)
ETCCDI_precip_malla.cdd(
    ARCHIVO_PR,
    ARCHIVO_SALIDA
)
```
