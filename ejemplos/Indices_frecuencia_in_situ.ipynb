## Índices ETCCDI de Precipitación-  
##  Frecuencia

| ID | Nombre | Definición | Unidades |
|----|--------|------------|----------|
|**R10mm**| Número de días de precipitación intensa | Conteo anual o estacional cuando PR ≥ 10 mm/año | días |
|**R20mm**| Número de días de precipitación muy intensa | Conteo anual o estacional cuando con PR ≥ 20 mm/año | días |


## Datos In Situ -`ETCCDI_precip_in_situ`

Las funciones para calcular los índices Categoria Frecuencia
Trabajan con datos de **estaciones meteorológicas** en formato `.txt`.

- Entrada: Un archivo .txt 

**Formato del archivo `.txt` de entrada:**
```
Year  Month  Day   PR     TX     TN
1991    1     1   -99.9   25.3   12.1
1991    1     2    0.0    26.1   13.4
...
```
Columnas separadas por espacios\
`PR`: precipitación (mm), `TX`: temp. máxima (°C), `TN`: temp. mínima (°C)\
Valor faltante: `-99.9`


- Ruta de salida: Donde el índice calculado se exporta como archivo `.xlsx`.\
**Formato esperado del archivo de salida:** \
```
Year   Índice
